### 8. Resolver el miniproyecto

Los datos siguientes son **ficticios** y se definen aquí. No provienen de ninguna empresa ni necesitan un archivo externo. Cada diccionario representa una venta y sus campos.

```python
ventas = [
    {"producto": "Cuaderno", "cantidad": 3, "precio": 4.0},
    {"producto": "Bolígrafo", "cantidad": 2, "precio": 1.5},
    {"producto": "Carpeta", "cantidad": 1, "precio": 5.0},
]
```

### 9. Dar nombre al cálculo

Define una función y úsala desde el bucle. `return` entrega el importe; `print` muestra los mensajes. Recorremos tres registros con cantidades y precios válidos para concentrarnos en el flujo de ejecución.

```python
def calcular_importe(cantidad, precio):
    return cantidad * precio


total_unidades = 0
total_euros = 0
for venta in ventas:
    importe_venta = calcular_importe(venta["cantidad"], venta["precio"])
    print(f"{venta['producto']}: {importe_venta:.2f} euros")
    total_unidades = total_unidades + venta["cantidad"]
    total_euros = total_euros + importe_venta

print(f"Unidades: {total_unidades}")
print(f"Total: {total_euros:.2f} euros")
```

## Comprobaciones

Para los datos originales deben salir `12`, `3` y `5` euros por venta: suman `20` euros y `6` unidades. Los `assert` siguientes comprueban esas expectativas. Si modificas los datos deliberadamente, revisa también las expectativas.

```python
assert total_unidades == 6
assert total_euros == 20
print("Comprobación correcta: 6 unidades y 20 euros")
```
